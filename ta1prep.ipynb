# 1. Fibonacci Numbers using Recursion and Iteration

## Problem Statement

To generate the nth Fibonacci numbers, evaluate its time and space
complexity by using:

(i) Recursion  
(ii) Iteration

Compare the time and space complexity of both the designs.

## Introduction

The Fibonacci sequence is a sequence in which every number is the sum
of the previous two numbers.

The sequence starts as:

0, 1, 1, 2, 3, 5, 8, 13, ...

The formula is:

F(n) = F(n-1) + F(n-2)

with:

F(0) = 0  
F(1) = 1

We implement Fibonacci in two ways:

1. Recursive approach
2. Iterative approach

## Working of Recursive Method

The function calls itself for the two previous Fibonacci numbers.

For example:

F(5)
= F(4) + F(3)
= (F(3) + F(2)) + (F(2) + F(1))

The same values are calculated many times. Therefore, recursion becomes
slow for larger values of n.

## Working of Iterative Method

The iterative method maintains two variables:

a = current Fibonacci number  
b = next Fibonacci number

The values are updated repeatedly inside a loop.

This avoids the repeated calculations performed by recursion.

## Complexity

### Recursive

Time Complexity: O(2^n)

The function makes two recursive calls for most values of n.

Space Complexity: O(n)

The recursive calls are stored in the function call stack.

### Iterative

Time Complexity: O(n)

The loop runs n times.

Space Complexity: O(1)

Only a constant number of variables are used.

## Comparison

| Method | Time Complexity | Space Complexity |
|--------|-----------------|------------------|
| Recursion | O(2^n) | O(n) |
| Iteration | O(n) | O(1) |

Therefore, the iterative method is much more efficient for generating
Fibonacci numbers.

```python
# Fibonacci using Recursion

def fibonacci_recursive(n):
    # Base case:
    # The 0th Fibonacci number is 0
    # The 1st Fibonacci number is 1
    if n <= 1:
        return n

    # Recursive formula:
    # F(n) = F(n-1) + F(n-2)
    return fibonacci_recursive(n - 1) + fibonacci_recursive(n - 2)


# Take input from the user
#n = 10  # Instead of: n = int(input("Enter n: "))
n = int(input("Enter n: "))

# Generate and display the nth Fibonacci number
print("Fibonacci number:", fibonacci_recursive(n))
```

```python
# Fibonacci using Iteration

n = int(input("Enter n: "))

# First two Fibonacci values
a = 0
b = 1

# Repeat n times
for i in range(n):
    # Display the current Fibonacci number
    print(a, end=" ")

    # Move to the next two values
    a, b = b, a + b
```

```python
# Comparison of Recursive and Iterative Fibonacci

n = int(input("Enter n: "))

# Recursive result
recursive_result = fibonacci_recursive(n)

# Iterative result
a = 0
b = 1

for i in range(n):
    a, b = b, a + b

iterative_result = a

print("Recursive result:", recursive_result)
print("Iterative result:", iterative_result)
```

# 2. Binary Search using Recursion and Iteration

## Problem Statement

To use binary search to quickly locate a book record in a library system
based on ISBN (0-100) or Title (A-Z).

Implement the search using:

(i) Recursion  
(ii) Iteration

Compare the time and space complexity of both the designs.

## Introduction

Binary Search is an efficient searching algorithm used on a sorted list.

Instead of checking every element one by one, binary search checks the
middle element and eliminates half of the remaining elements after
every comparison.

For example, if we search for 70 in:

10, 20, 30, 40, 50, 60, 70, 80, 90

We first check 50.

Since 70 is greater than 50, we ignore the left half.

We then search only:

60, 70, 80, 90

This process continues until the element is found or the search space
becomes empty.

## Important Requirement

Binary Search works only when the data is sorted.

## Working

1. Find the middle element.
2. Compare the middle element with the search value.
3. If both are equal, the element is found.
4. If the search value is smaller, search the left half.
5. If the search value is larger, search the right half.
6. Repeat until the element is found.

## Complexity

Best Case: O(1)

The element is found at the middle during the first comparison.

Average Case: O(log n)

Worst Case: O(log n)

The search space is divided by two in every step.

### Recursive Space

O(log n), because recursive function calls are stored in the call stack.

### Iterative Space

O(1), because only a few variables are required.

| Method | Time | Space |
|--------|------|-------|
| Recursive | O(log n) | O(log n) |
| Iterative | O(log n) | O(1) |

```python
# Binary Search using Recursion

def binary_search_recursive(arr, low, high, key):

    # If the search range becomes empty,
    # the element does not exist
    if low > high:
        return -1

    # Find the middle position
    mid = (low + high) // 2

    # If the middle element is the key, return its position
    if arr[mid] == key:
        return mid

    # If key is smaller, search the left half
    elif key < arr[mid]:
        return binary_search_recursive(arr, low, mid - 1, key)

    # Otherwise search the right half
    else:
        return binary_search_recursive(arr, mid + 1, high, key)


# Input elements
arr = list(map(int, input("Enter sorted ISBN numbers: ").split()))

# Input search value
key = int(input("Enter ISBN to search: "))

# Perform binary search
result = binary_search_recursive(arr, 0, len(arr) - 1, key)

# Display result
if result == -1:
    print("ISBN not found")
else:
    print("ISBN found at position:", result)
```

```python
# Binary Search using Iteration

arr = list(map(int, input("Enter sorted ISBN numbers: ").split()))
key = int(input("Enter ISBN to search: "))

# Start and end of the search range
low = 0
high = len(arr) - 1

found = -1

# Continue while there are elements to search
while low <= high:

    # Find the middle position
    mid = (low + high) // 2

    # Element found
    if arr[mid] == key:
        found = mid
        break

    # Search right half
    elif key > arr[mid]:
        low = mid + 1

    # Search left half
    else:
        high = mid - 1


# Display result
if found == -1:
    print("ISBN not found")
else:
    print("ISBN found at position:", found)
```

# 3. Quick Sort using Divide and Conquer

## Problem Statement

Sort a given set of n integer elements using Sort method and compute its
time complexity. Run the program for varied values of n and record the
time taken to sort. The elements can be read from a user or can be
generated using the random number generator and pivot element.

Demonstrate how the divide and conquer method works along with its time
complexity analysis:

- Worst Case
- Average Case
- Best Case

## Introduction

Quick Sort is a Divide and Conquer sorting algorithm.

It works by selecting one element as a pivot and dividing the array into
smaller parts based on the pivot.

Elements smaller than the pivot are placed on one side and elements
larger than the pivot are placed on the other side.

The same process is then applied recursively to the two smaller parts.

## Working

For example:

[8, 3, 7, 4, 9, 2]

Choose 8 as pivot.

Smaller elements:
[3, 7, 4, 2]

Larger elements:
[9]

Now sort the smaller and larger parts recursively.

Finally, combine:

sorted left + pivot + sorted right

This is the Divide and Conquer approach.

## Time Complexity

### Best Case: O(n log n)

The pivot divides the array into two approximately equal parts.

### Average Case: O(n log n)

On average, the partitions are reasonably balanced.

### Worst Case: O(n²)

If the pivot is repeatedly the smallest or largest element, one side
contains almost all elements.

For example, this can happen when the input is already sorted and the
first or last element is always selected as pivot.

## Space Complexity

The recursive calls require stack space.

Average/Best: O(log n)

Worst: O(n)

```python
# Quick Sort using Divide and Conquer

import random
import time


def quick_sort(arr):

    # If the array has 0 or 1 element, it is already sorted
    if len(arr) <= 1:
        return arr

    # Select the first element as pivot
    pivot = arr[0]

    # Store elements smaller than pivot
    left = []

    # Store elements equal to pivot
    middle = []

    # Store elements larger than pivot
    right = []

    # Divide the array around the pivot
    for x in arr:

        if x < pivot:
            left.append(x)

        elif x == pivot:
            middle.append(x)

        else:
            right.append(x)

    # Recursively sort left and right parts
    return quick_sort(left) + middle + quick_sort(right)


# Generate different input sizes
sizes = [10, 100, 500, 1000]

for n in sizes:

    # Generate random numbers
    arr = [random.randint(1, 1000) for i in range(n)]

    # Start time
    start = time.time()

    # Sort the array
    sorted_arr = quick_sort(arr)

    # End time
    end = time.time()

    print("Number of elements:", n)
    print("Time taken:", end - start, "seconds")
    print()
```

## Quick Sort Divide and Conquer Structure

The process can be represented as:

Original Array

        ↓
     Choose Pivot

        ↓
   Divide the Array

     /         \
 Smaller      Larger

    ↓            ↓

Sort Again    Sort Again

     \          /

        Combine

           ↓

     Sorted Array

The efficiency of Quick Sort depends strongly on how well the pivot
divides the array.

# 4. Merge Sort for Examination Enrolment Numbers

## Problem Statement

Sort examination answer sheets of individual enrolment numbers in a
given ordered by dean of examination. Identify sorting algorithm for
same and discuss time and space complexity.

## Algorithm Selection

The suitable algorithm is Merge Sort.

Merge Sort is a Divide and Conquer sorting algorithm that repeatedly
divides the list into two halves and then merges the sorted halves.

## Working

Suppose the enrolment numbers are:

[45, 12, 78, 23, 10, 56]

First, divide:

[45, 12, 78]    [23, 10, 56]

Continue dividing until individual elements remain.

Then merge the elements in sorted order.

The process is:

Divide → Divide → Sort/Merge → Combine

## Why Merge Sort?

Merge Sort has a guaranteed O(n log n) time complexity.

This means even in the worst case, its performance remains efficient.

## Complexity

Best Case: O(n log n)

Average Case: O(n log n)

Worst Case: O(n log n)

Space Complexity: O(n)

Additional memory is needed while merging the arrays.

```python
# Merge Sort

def merge_sort(arr):

    # If there is only one element,
    # the array is already sorted
    if len(arr) <= 1:
        return arr

    # Find the middle
    mid = len(arr) // 2

    # Divide the array into two halves
    left = merge_sort(arr[:mid])
    right = merge_sort(arr[mid:])

    # Merge the two sorted halves
    return merge(left, right)


def merge(left, right):

    result = []

    i = 0
    j = 0

    # Compare elements from both arrays
    while i < len(left) and j < len(right):

        if left[i] < right[j]:
            result.append(left[i])
            i += 1

        else:
            result.append(right[j])
            j += 1

    # Add remaining elements from left array
    while i < len(left):
        result.append(left[i])
        i += 1

    # Add remaining elements from right array
    while j < len(right):
        result.append(right[j])
        j += 1

    return result


# Input enrolment numbers
arr = list(map(int, input("Enter enrolment numbers: ").split()))

print("Before sorting:", arr)

# Sort the enrolment numbers
arr = merge_sort(arr)

print("After sorting:", arr)
```

# 5. Matrix Multiplication for Augmented Reality Transformations

## Problem Statement

Accelerating Image Transformations in Augmented Reality

A mobile AR app applies various transformations to object models using
large matrices. However, with increasing model sizes and device
resolution, users have reported noticeable lags during rendering
transitions. The development team is looking for an approach that
reduces the computational load without compromising transformation
accuracy. Examine how transformations are applied, identify
opportunities for optimization in matrix operations, and propose an
efficient technique that scales better on low-power devices.

## Introduction

In Augmented Reality, objects are transformed using mathematical
matrices.

Common transformations include:

- Translation
- Rotation
- Scaling

A transformation can be represented using a matrix.

When several transformations are applied, matrix multiplication is used
to combine them.

For example:

Object Coordinates
        ↓
Transformation Matrix
        ↓
Matrix Multiplication
        ↓
Transformed Coordinates
        ↓
Rendered AR Object

## Matrix Multiplication

For two matrices A and B:

C = A × B

Each element of C is calculated by multiplying corresponding elements
and adding the results.

For n × n matrices, the standard method uses three nested loops.

## Complexity

For two n × n matrices:

Time Complexity: O(n³)

Space Complexity: O(n²)

The cubic time complexity can become expensive for large matrices.

## Optimization

For large AR systems and low-power devices, performance can be improved
by using optimized matrix libraries, GPU acceleration, or specialized
matrix multiplication techniques.

One important theoretical alternative is Strassen's matrix
multiplication algorithm, which reduces the asymptotic complexity from:

O(n³)

to approximately:

O(n^2.81)

However, for small transformation matrices commonly used in graphics,
standard optimized matrix operations are generally more practical.

```python
# Matrix Multiplication

# Function to multiply two matrices
def multiply(A, B):

    # Number of rows and columns
    rows = len(A)
    cols = len(B[0])

    # Create result matrix filled with zeros
    C = [[0] * cols for i in range(rows)]

    # Three loops perform matrix multiplication
    for i in range(rows):
        for j in range(cols):
            for k in range(len(B)):
                C[i][j] = C[i][j] + A[i][k] * B[k][j]

    return C


# Example matrices
A = [
    [1, 2],
    [3, 4]
]

B = [
    [5, 6],
    [7, 8]
]

# Multiply matrices
C = multiply(A, B)

print("Result of Matrix Multiplication:")

# Display result
for row in C:
    print(row)
```

## Optimization Idea

The standard matrix multiplication algorithm is easy to implement but
requires O(n³) operations for n × n matrices.

For applications involving large matrices, better performance can be
obtained using:

1. Optimized numerical libraries such as BLAS.
2. GPU-based matrix operations.
3. Parallel matrix multiplication.
4. Strassen's algorithm for sufficiently large matrices.

For a mobile AR application, the practical choice is usually an
optimized matrix library or hardware acceleration rather than manually
implementing Strassen's algorithm.

# 6. Fractional Knapsack using Greedy Method

## Problem Statement

An Optimal Cargo Loading delivery company has to load a truck with items
of different weights and profits. The truck can carry a limited total
weight. Implement the problem using a Greedy approach to maximize
profit. The program should allow fractional inclusion of items and
display which items (or fractions) were selected and the resulting
maximum profit.

## Introduction

In the Fractional Knapsack problem, each item has:

- Weight
- Profit

The truck has a fixed weight capacity.

Unlike the 0/1 Knapsack problem, an item can be partially selected.

For example, if an item weighs 10 kg and only 5 kg capacity remains, we
can take 50% of that item.

## Greedy Strategy

For every item, calculate:

Profit / Weight

This gives the profit earned per unit of weight.

The item with the highest profit-to-weight ratio is selected first.

Then continue selecting items in decreasing order of this ratio.

## Working

Suppose:

Item 1: weight = 10, profit = 60  
Item 2: weight = 20, profit = 100

Ratios:

Item 1 = 60 / 10 = 6  
Item 2 = 100 / 20 = 5

Therefore, Item 1 is selected first.

If there is insufficient capacity to take the complete next item,
only the required fraction is selected.

## Complexity

Sorting the items based on profit/weight ratio:

O(n log n)

Selecting the items:

O(n)

Therefore:

Overall Time Complexity: O(n log n)

Space Complexity: O(n)

## Important Point

The Greedy method gives the optimal answer for the Fractional Knapsack
problem because fractions of items are allowed.

```python
# Fractional Knapsack using Greedy Method

# Number of items
n = int(input("Enter number of items: "))

items = []

# Read weight and profit of each item
for i in range(n):

    weight = float(input("Enter weight of item " + str(i + 1) + ": "))
    profit = float(input("Enter profit of item " + str(i + 1) + ": "))

    # Calculate profit per unit weight
    ratio = profit / weight

    # Store item number, weight, profit and ratio
    items.append([i + 1, weight, profit, ratio])


# Truck capacity
capacity = float(input("Enter truck capacity: "))

# Sort items by profit/weight ratio
# Highest ratio should come first
items.sort(key=lambda x: x[3], reverse=True)

total_profit = 0

print("\nSelected Items:")

# Select items greedily
for item in items:

    number = item[0]
    weight = item[1]
    profit = item[2]

    # If the complete item can fit
    if weight <= capacity:

        capacity = capacity - weight
        total_profit = total_profit + profit

        print("Item", number, ": 100% selected")

    # Otherwise take only the required fraction
    else:

        fraction = capacity / weight

        total_profit = total_profit + profit * fraction

        print("Item", number, ":", fraction * 100, "% selected")

        # Truck is now full
        capacity = 0

        break


print("\nMaximum Profit:", total_profit)
```
